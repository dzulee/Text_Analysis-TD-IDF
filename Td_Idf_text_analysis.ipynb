```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from nltk.tokenize import word_tokenize
import nltk
nltk.download('punkt_tab')
```

# We import neccessary Scikit-learn libraries for **keyword extraction**
we follow a standard **Natural Language Processing** ***(NLP) ***pipeline.
Using scikit-learn for the math (TF-IDF) and standard Python for the processing steps, we will follow this 4-Step Process to Keyword Selection


*   Preprocessing: Convert text to lowercase and remove punctuation so that "Hotel" and "hotel" are treated as the same word.
*   Stop Word Removal: Filter out "noise" words (the, is, at, for) that don't carry meaning.
*   Tokenization: Break the paragraph into individual word units.
*   TF-IDF Weighting: Calculate which words are "significant." A word is significant if it appears often in this text but isn't a common word in general English.
















```python

# Loading PunktSentenceTokenizer using English pickle file
tokenizer = nltk.data.load('tokenizers/punkt/PY3/english.pickle')
text=tokenizer.tokenize(text)
print(text)
```

```python
# 2. Initialize the TF-IDF Vectorizer
# 'english' stop_words removes common words like 'the', 'and', 'is'
vectorizer = TfidfVectorizer(stop_words='english', ngram_range=(1,2), lowercase=True)
```

```python


# 3. Process the text
tfidf_matrix = vectorizer.fit_transform(text)
feature_names = vectorizer.get_feature_names_out()
dense = tfidf_matrix.todense()
denselist = dense.tolist()

```

```python
# 5. Create a sorted DataFrame
keywords_df =  pd.DataFrame(denselist, columns=feature_names)
keywords_df = keywords_df.mean().sort_values(ascending=False)
```

```python
# Display Top 10 Keywords
print("Top Keywords identified by TF-IDF:")
print(keywords_df.head(20))
```
